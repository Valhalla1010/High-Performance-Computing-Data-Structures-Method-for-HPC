```python
import numpy as np


def matrices_array(n):
    A = np.random.random((n,n)).astype(np.float64)
    B = np.random.random((n,n)).astype(np.float64)
    C= np.zeros_like(A)
    return A, B, C

def dgemm(A, B, C, n):
    for i in range(n):
        for j in range(n):
            for k in range(n):
                C[i][j] += A[i][k] * B[k][j]
    return C



n = 5
A, B, C = matrices_array(n)
print(f"A:\n{A}")
print(f"B:\n{B}")

C_result = dgemm(A, B, C, n)
print(f"C:\n{C_result}")
```

Small matrices (n < 10), excution times are stable (low std) since operations fit in CPU cash. 
Large matrices (n> 100), standard deviation increases due to cache misses, memory access overheads.
NumPy is the best choice in Average, min, max and Std time for performance and stability as it minimiez memory overheads. But in small size on Standard deviation show that list is good.  

The total number of operations for DGEMM is 2N^3. My CPU has clock speed= 1.90 GHz = 1.90 * 2^30 = 2,040,109,466.
For large matrices (N = 1000), dgemm_numpy is nearly 10,000x faster than dgemm_list or dgemm_array, and numpy achieves a performance far beyond the theoretical peak of the CPU. Measured FLOPS values show that numpy's performance increases with larger matrices, whereas the list and array implementations remain bottlenecked by poor optimization.
